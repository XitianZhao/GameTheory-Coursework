```python
import numpy as np
import matplotlib.pyplot as plt


# Parameters
s = 0.2
t = 0



theta = 30.0 / 360.0 * 2.0 * np.pi  # 30 degrees in radians
proj = np.array([
    [-np.cos(theta),  np.cos(theta),  0.0],
    [-np.sin(theta), -np.sin(theta),  1.0]
])
# So the vertices e1,e2,e3 map to:
# e1 -> (-cosθ, -sinθ), e2 -> (cosθ, -sinθ), e3 -> (0,1)


def payoff_BH(p, s, t):
    # p = (p0, p1, p2) = (AH, AL, T)
    p0, p1, p2 = p
    # minus sign: defender minimizes attacker payoff
    return -(0 * p0 + (109 + 100 * s) * p1 + (74 - 283 * t) * p2)

def payoff_BL(p, s, t):
    p0, p1, p2 = p
    return -((32 + 78 * s) * p0 + (-27) * p1 + (74 - 283 * t) * p2)

def best_response(p):
    bh = payoff_BH(p, s, t)
    bl = payoff_BL(p, s, t)
    return 1 if bh > bl else 0  # 1 = BH best, 0 = BL best


# Boundary line in simplex: BH = BL
# Equation: (136 + 100 s) p1 - (32 + 78 s) p0 = 0
# => p1 = R * p0, with p2 = 1 - p0 - p1 >= 0

def boundary_points(s, n=300):
    A = 136 + 100 * s
    B = 32 + 78 * s
    R = B / A  # slope p1 = R * p0

    pts = []
    # p2 >= 0 ⇒ 1 - p0 - R p0 >= 0 ⇒ p0 <= 1/(1+R)
    max_p0 = 1.0 / (1.0 + R)

    for p0 in np.linspace(0, max_p0, n):
        p1 = R * p0
        p2 = 1.0 - p0 - p1
        if p1 >= 0 and p2 >= 0:
            pts.append(np.array([p0, p1, p2]))
    return np.array(pts)

boundary = boundary_points(s)

# Take the intersection on the base edge (last point)
intersection_prob = boundary[-1]  # already (p0,p1,p2)
print("Intersection on boundary (probabilities):")
print(f"p0 = {intersection_prob[0]:.4f}, p1 = {intersection_prob[1]:.4f}, p2 = {intersection_prob[2]:.4f}")


# Build a grid over the simplex for BR regions
res = 200
p0_list, p1_list, p2_list = [], [], []
vals = []

for i in np.linspace(0, 1, res):
    for j in np.linspace(0, 1 - i, res):
        k = 1.0 - i - j
        p = np.array([i, j, k])
        p0_list.append(i)
        p1_list.append(j)
        p2_list.append(k)
        vals.append(best_response(p))

p0_arr = np.array(p0_list)
p1_arr = np.array(p1_list)
p2_arr = np.array(p2_list)
vals_arr = np.array(vals)


# Project BR grid
pts_simplex = np.vstack([p0_arr, p1_arr, p2_arr])  # shape (3, N)
pts_2d = proj @ pts_simplex                         # shape (2, N)
x = pts_2d[0]
y = pts_2d[1]

# Project boundary
boundary_2d = proj @ boundary.T
bx, by = boundary_2d[0], boundary_2d[1]

# Project intersection point
intersection_2d = proj @ intersection_prob
ix, iy = intersection_2d[0], intersection_2d[1]

# Project the simplex edges for drawing the triangle
ts = np.linspace(0, 1, 1000)
PBd1 = proj @ np.array([ts, 1 - ts, 0 * ts])  # edge e1-e2
PBd2 = proj @ np.array([0 * ts, ts, 1 - ts])  # edge e2-e3
PBd3 = proj @ np.array([ts, 0 * ts, 1 - ts])  # edge e1-e3


plt.figure(figsize=(7, 7))
plt.box(False)
plt.axis("equal")
plt.axis("off")


plt.scatter(x, y, c=vals_arr, cmap="Greys_r", s=5)

# Boundary line BH = BL
plt.plot(bx, by, color="red", linewidth=2)

# Intersection point on boundary
plt.plot(ix, iy, "o", color="white", markersize=8)
plt.text(ix + 0.03, iy,
         f"({intersection_prob[0]:.2f}, {intersection_prob[1]:.2f}, {intersection_prob[2]:.2f})",
         fontsize=9, color="white")

# Triangle boundaries
plt.plot(PBd1[0], PBd1[1], ".", color="black", markersize=2)
plt.plot(PBd2[0], PBd2[1], ".", color="black", markersize=2)
plt.plot(PBd3[0], PBd3[1], ".", color="black", markersize=2)

# Label vertices: e1 = AH, e2 = AL, e3 = T
e1 = proj @ np.array([1., 0., 0.])
e2 = proj @ np.array([0., 1., 0.])
e3 = proj @ np.array([0., 0., 1.])

plt.text(e1[0] - 0.12, e1[1] - 0.05, "AH", fontsize=12)
plt.text(e2[0] + 0.05, e2[1] - 0.05, "AL", fontsize=12)
plt.text(e3[0] - 0.03, e3[1] + 0.08, "T", fontsize=12)

# --- Region label positions in simplex coords ---
p_BL = np.array([0.10, 0.80, 0.10])   # mostly AL → BH region
p_BH = np.array([0.80, 0.10, 0.10])   # mostly AH → BL region

# --- Project to 2D ---
x_BH, y_BH = proj @ p_BH
x_BL, y_BL = proj @ p_BL

# --- Label them ---
plt.text(
    x_BH, y_BH,
    "BH best response",
    ha='center', va='center',
    color="red", fontsize=12
)

plt.text(
    x_BL, y_BL,
    "BL best response",
    ha='center', va='center',
    color="red", fontsize=12
)
plt.title("Defender Best Response Regions with s=0.2, t=0")

plt.show()
```

```python
# Iterations based on regret minimisation

import numpy as np


s_true = 0.2
t_true = 0


U = np.array([
    [0, 32 + 78*s_true],            # AH vs (BH,BL)
    [109 + 100*s_true, -27],        # AL
    [74 - 283*t_true, 74 - 283*t_true]   # T
])

U_def = -U  # zero-sum assumption



A_actions = 3
D_actions = 2
T_iter = 10000

regret_A = np.zeros(A_actions)
regret_D = np.zeros(D_actions)

strat_sum_A = np.zeros(A_actions)
strat_sum_D = np.zeros(D_actions)

# storage of observed data for s,t estimation
observed_AL_BH = []  # AL vs BH payoff → gives 109 + 100s
observed_AH_BL = []  # AH vs BL payoff → gives 32 + 78s
observed_T_any = []  # T vs BH or BL payoff → gives 74 - 283t


def regret_matching(regrets):
    pos = np.maximum(regrets, 0)
    s = pos.sum()
    if s > 0:
        return pos / s
    return np.ones_like(pos) / len(pos)


def estimate_parameters():
    est_s = None
    est_t = None
    
    # Estimate s from AL-BH or AH-BL observations
    s_vals = []
    
    # AL vs BH gives payoff = 109 + 100 s → s = (payoff - 109)/100
    if len(observed_AL_BH) > 0:
        for payoff in observed_AL_BH:
            s_vals.append((payoff - 109) / 100)
    
    # AH vs BL gives payoff = 32 + 78 s → s = (payoff - 32)/78
    if len(observed_AH_BL) > 0:
        for payoff in observed_AH_BL:
            s_vals.append((payoff - 32) / 78)
    
    if len(s_vals) > 0:
        est_s = np.mean(s_vals)
    
    # Estimate t from T matches: payoff = 74 - 283t → t = (74 - payoff)/283
    if len(observed_T_any) > 0:
        est_t = np.mean([(74 - payoff)/283 for payoff in observed_T_any])
    
    return est_s, est_t



# Main regret-matching loop

for t in range(1, T_iter + 1):

    sigma_A = regret_matching(regret_A)
    sigma_D = regret_matching(regret_D)

    strat_sum_A += sigma_A
    strat_sum_D += sigma_D

    # Sample action profile
    a = np.random.choice(A_actions, p=sigma_A)
    d = np.random.choice(D_actions, p=sigma_D)

    uA = U[a, d]
    uD = U_def[a, d]

    # Record data for parameter estimation
    if a == 1 and d == 0:   # AL vs BH
        observed_AL_BH.append(uA)
    if a == 0 and d == 1:   # AH vs BL
        observed_AH_BL.append(uA)
    if a == 2:             # T vs anything
        observed_T_any.append(uA)

    # Update attacker regrets
    for a_alt in range(A_actions):
        regret_A[a_alt] += U[a_alt, d] - uA

    # Update defender regrets
    for d_alt in range(D_actions):
        regret_D[d_alt] += U_def[a, d_alt] - uD

    # Print progress every 100 rounds

    if t % 100 == 0:
        est_s, est_t = estimate_parameters()
        print(f"\n=== Round {t} ===")
        print(f"Attacker strategy: AH={sigma_A[0]:.3f}, AL={sigma_A[1]:.3f}, T={sigma_A[2]:.3f}")
        print(f"Defender strategy: BH={sigma_D[0]:.3f}, BL={sigma_D[1]:.3f}")
        print(f"Estimated s: {est_s}")
        print(f"Estimated t: {est_t}")

# Final average strategies
avg_A = strat_sum_A / T_iter
avg_D = strat_sum_D / T_iter



est_s, est_t = estimate_parameters()
print("\n=== FINAL PARAMETER ESTIMATES ===")
print("Estimated s:", est_s)
print("Estimated t:", est_t)

print("\n=== FINAL AVERAGE STRATEGIES ===")
print(f"Attacker: AH={avg_A[0]:.3f}, AL={avg_A[1]:.3f}, T={avg_A[2]:.3f}")
print(f"Defender: BH={avg_D[0]:.3f}, BL={avg_D[1]:.3f}")
```
